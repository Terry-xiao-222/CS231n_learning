## Inline Question 1:
Did you notice anything about the comparative difficulty of training the three-layer network vs. training the five-layer network? In particular, based on your experience, which network seemed more sensitive to the initialization scale? Why do you think that is the case?

## Answer:

五层网络比三层网络更难训练，并且对权重初始化尺度 `weight_scale` 更敏感。

这是因为在深层网络中，前向传播的信号和反向传播的梯度都要经过更多次矩阵乘法。如果初始化权重过小，激活值和梯度会随着层数增加而逐渐减小，导致梯度消失，网络学习缓慢；如果权重过大，激活值和梯度可能逐层放大，导致梯度爆炸和训练不稳定。此外，不合适的初始化还可能使更多 ReLU 神经元长期输出零。

三层网络的传播路径较短，因此信号和梯度的缩放问题没有五层网络严重，对初始化尺度的容忍范围也更大。五层网络则需要更合适的权重初始化，才能使各层激活值和梯度保持在合理范围内。

## Inline Question 2:

AdaGrad, like Adam, is a per-parameter optimization method that uses the following update rule:

```
cache += dw**2
w += - learning_rate * dw / (np.sqrt(cache) + eps)
```

John notices that when he was training a network with AdaGrad that the updates became very small, and that his network was learning slowly. Using your knowledge of the AdaGrad update rule, why do you think the updates would become very small? Would Adam have the same issue?


## Answer:
AdaGrad 会不断累加过去所有梯度的平方。由于 cache 只会增加而不会减小，训练时间越长，sqrt(cache) 就越大。

AdaGrad 的参数更新量约等于：

learning_rate * dw / (sqrt(cache) + eps)

因此，当 cache 不断增大时，分母会越来越大，相当于每个参数的有效学习率不断减小。即使当前梯度仍然不小，参数的更新量也可能变得非常小，导致网络学习速度越来越慢，甚至过早停止学习。

Adam 通常不会出现同样严重的问题。Adam 使用梯度和梯度平方的指数移动平均，较早的梯度会随着训练逐渐被遗忘，因此其二阶矩估计不会像 AdaGrad 的 cache 那样一直无限累加。Adam 的有效学习率通常不会持续减小到接近零。不过，当网络接近最优解、梯度本身变小时，Adam 的参数更新量也会自然变小。




# Train a Good Model!
Train the best fully connected model that you can on CIFAR-10, storing your best model in the `best_model` variable. We require you to get at least 50% accuracy on the validation set using a fully connected network.

If you are careful it should be possible to get accuracies above 55%, but we don't require it for this part and won't assign extra credit for doing so.
````python
best_model = None
best_solver = None
best_val_acc = -1.0
experiment_results = []
experiment_solvers = {}

normalization_methods = [
    "batchnorm",
    "layernorm",
]

optimizer_configs = {
    "sgd_momentum": {
        "learning_rate": 1e-2,
        "momentum": 0.9,
    },
    "rmsprop": {
        "learning_rate": 1e-3,
        "decay_rate": 0.99,
        "epsilon": 1e-8,
    },
    "adam": {
        "learning_rate": 1e-3,
        "beta1": 0.9,
        "beta2": 0.999,
        "epsilon": 1e-8,
    },
}

run_id = 0
total_runs = len(normalization_methods) * len(optimizer_configs)

for normalization in normalization_methods:
    for update_rule, optim_config in optimizer_configs.items():
        run_id += 1
        np.random.seed(231 + run_id)

        print("=" * 70)
        print(
            f"Experiment {run_id}/{total_runs}: "
            f"normalization={normalization}, "
            f"optimizer={update_rule}"
        )

        model = FullyConnectedNet(
            hidden_dims=[256, 128],
            input_dim=3 * 32 * 32,
            num_classes=10,
            normalization=normalization,
            dropout_keep_ratio=1,
            reg=1e-3,
            weight_scale=1e-2,
            dtype=np.float32,
        )

        solver = Solver(
            model,
            data,
            update_rule=update_rule,
            optim_config=optim_config.copy(),
            lr_decay=0.95,
            num_epochs=8,
            batch_size=200,
            print_every=200,
            verbose=True,
        )

        solver.train()

        experiment_name = f"{normalization}_{update_rule}"
        experiment_solvers[experiment_name] = solver

        result = {
            "normalization": normalization,
            "optimizer": update_rule,
            "best_val_acc": solver.best_val_acc,
            "final_loss": solver.loss_history[-1],
        }
        experiment_results.append(result)

        print(
            f"{experiment_name}: "
            f"best validation accuracy = {solver.best_val_acc:.4f}"
        )

        if solver.best_val_acc > best_val_acc:
            best_val_acc = solver.best_val_acc
            best_model = solver.model
            best_solver = solver

print("\n" + "=" * 70)
print("Experiment summary")

experiment_results.sort(
    key=lambda result: result["best_val_acc"],
    reverse=True,
)

for result in experiment_results:
    print(
        f"{result['normalization']:10s} | "
        f"{result['optimizer']:12s} | "
        f"val_acc={result['best_val_acc']:.4f} | "
        f"final_loss={result['final_loss']:.4f}"
    )

print("=" * 70)
print("Best validation accuracy:", best_val_acc)
print(
    "Best configuration:",
    experiment_results[0]["normalization"],
    "+",
    experiment_results[0]["optimizer"],
)
````
