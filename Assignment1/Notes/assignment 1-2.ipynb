## 1. np.random.randn()

生成服从**标准正态分布**的随机数：均值$\mu=0$，方差$\sigma^2=1$，$\mathcal N(0,1)$

```python
import numpy as np

# 1个标量
print(np.random.randn())  

# 一维数组 shape=(4,)
a = np.random.randn(4)
print(a) # (4,)

# 二维数组 shape=(2,3)
b = np.random.randn(2,3)
print(b) # (2, 3)
```

## 3. np.hstack((a,b)) & np.vstack((a,b))

horizontal stack 水平拼接，按列方向把数组拼在一起（左右拼）

vertical stack 垂直拼接，按行方向把数组拼在一起（上下拼）

```python
a = np.array([[1,2],[3,4]])   #(2,2)
b = np.array([[5,6],[7,8]])   #(2,2)
c = np.hstack((a,b))
print(c) 
print(np.concatenate((a, b), axis=1))
```

```python
a = np.array([[1,2],[3,4]])   #(2,2)
b = np.array([[5,6],[7,8]])   #(2,2)
c = np.vstack((a,b))
print(c) 
print(np.concatenate((a, b), axis=0))
```

```python
a = np.array([1, 2])   #(2,)
b = np.array([3, 4])   #(2,)
c = np.vstack((a, b))
d = np.hstack((a, b))
e = np.concatenate((a, b), axis=0)
## 注意np.concatenate对于两个一维数组没有axis=1，返回还是一维数组
print(c) # (2, 2)
print(d) # (4,)
print(e) # (4,)
```

## 5.np.max(a, axis, keepdims)

`axis=0`沿着**行方向求 max** → 对**每一列（每个样本）单独求最大值**

`axis=1`沿着**列方向求 max** → 对**每一行（每个样本）单独求最大值**

`keepdims=False`：`np.max(scores,axis=1)` → shape `(N,)` 不保留被压缩的轴

`keepdims=True`：`np.max(scores,axis=1,keepdims=True)` → shape **`(N, 1)`** 保留被压缩的轴

如要广播，必须`keepdims=True`

```python
import numpy as np
scores = np.array([
    [5, 1, 3],   #样本0，max=5
    [2, 7, 4]    #样本1，max=7
])

row_max = np.max(scores, axis=1, keepdims=True)
row_max2 = np.max(scores, axis=1)
print(row_max, row_max2)
'''
[[5]
 [7]]
'''

scores -= row_max
print(scores)
'''
[[ 0 -4 -2]
 [-5  0 -3]]
'''
```
