## 1. np.random.rand()

生成[0,1) 均匀分布

```python
import numpy as np
x = np.array([[1 ,2, 3], [4, 5, 6]])
print(np.random.rand(*x.shape)) # *是解包操作
```

## 2. np.random.randn()

生成标准正态分布（正负都有）

```python
print(np.random.randn(2,3))
```

## 3. Bool判断



```python
x = np.array([[1 ,2, 3], [4, 5, 6]])
y = np.random.rand(*x.shape)
r = y < 0.8
print(y)
print(r)
```

## 4. np.pad(array, pad_width, mode='constant')

给数组各个维度**前后填充数值**，mode='constant' 填充0

```python
arr = np.array([[1,2],[3,4]])
# shape (2,2)
# 第0维：行；第1维：列
p = ((0,1), (2,2))
out = np.pad(arr, p, mode='constant')
print(out)
```

## 5. np.prod(a, axis=None, keepdims=False)

计算数组元素的**乘积**（product），把数组里数字相乘

```python
import numpy as np

arr = np.array([2,3,4])
print(np.prod(arr))   # 2*3*4 = 24

arr2 = np.array([[2,3,4],[5,6,7]])
print(np.prod(arr2, axis=0)) 
print(np.prod(arr2, axis=1, keepdims=True)) 
```

## 6. np.linspace(start, stop, num)

**生成等间隔的一维数组**，在 `start ~ stop` 之间，均匀产出 `num` 个点（首尾都包含）

```python
print(np.linspace(-0.1, 0.5, 3))
```
