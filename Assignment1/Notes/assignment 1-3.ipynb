## 1. np.reshape()

```python
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([5, 4, 3, 2, 1, 0])
b_reshaped = b.reshape(a.shape)
print(b_reshaped)
```

## 2. np.maximum()

**np.maximum(x,y)**：两个输入，**对应位置比大小，取大的那个**，支持广播。二元运算。

```python
import numpy as np

a = np.array([1, 4, 3])
b = np.array([2, 2, 5])
c = np.array([[-1, 3, -6], [1, -3, 7]])
res1 = np.maximum(a, b)
res2 = np.maximum(c, 0)
print(res1)   # [2 4 5]
print(res2)   # [[0 3 0] [1 0 7]]
```

## 3. np.max(a, axis=?, keepdims=?

**一个数组**内部聚合求最大

`keepdims`在没有声明的情况下默认为False

```python
a = np.array([[-1, 3, -6], [1, -3, 7]])
print(np.max(a, axis=0)) 
print(np.max(a, axis=0, keepdims=True))   
print(np.max(a, axis=1)) 
print(np.max(a, axis=1, keepdims=True))   
```
